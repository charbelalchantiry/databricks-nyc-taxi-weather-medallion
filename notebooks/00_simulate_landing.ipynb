```python
from pyspark.sql import functions as F

base = "/Volumes/workspace/taxi_weather/raw"
df = spark.table("samples.nyctaxi.trips")

jan = df.filter(F.month("tpep_pickup_datetime") == 1)
feb = df.filter(F.month("tpep_pickup_datetime") == 2)

jan.write.mode("overwrite").parquet(f"{base}/landing/taxi/jan")
feb.write.mode("overwrite").parquet(f"{base}/staging/taxi_feb")

print("total:", df.count(), "| jan:", jan.count(), "| feb:", feb.count())
```
